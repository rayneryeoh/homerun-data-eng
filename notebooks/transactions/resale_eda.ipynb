```python
import pandas as pd
```

```python
resale_df = pd.read_csv("../../data/raw/transactions/ResaleflatpricesbasedonregistrationdatefromJan2017onwards.csv")
```

```python
resale_df
```

```python
resale_df["month"] = pd.to_datetime(resale_df["month"], format="%Y-%m").dt.to_period("M")
```

```python
# Convert month from string to monthly period
resale_df
```

```python
# Define the reference month (current month)
reference_month = pd.Period("2026-11", freq="M")
```

```python
# Latest 3 complete months: Aug-Oct 2026
lookback_months = 3
end_month = reference_month - 1
start_month = end_month - (lookback_months - 1)
```

```python
end_month
```

```python
start_month
```

```python
# Listings proxy: recent transactions
listings_df = resale_df.loc[
    resale_df["month"].between(start_month, end_month)
].copy()
```

```python
# Historical: all transactions before the proxy window
historical_df = resale_df.loc[
    resale_df["month"] < start_month
].copy()
```
